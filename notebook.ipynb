Every year, American high school students take SATs, which are standardized tests intended to measure literacy, numeracy, and writing skills. There are three sections - reading, math, and writing, each with a **maximum score of 800 points**. These tests are extremely important for students and colleges, as they play a pivotal role in the admissions process.

Analyzing the performance of schools is important for a variety of stakeholders, including policy and education professionals, researchers, government, and even parents considering which school their children should attend. 

You have been provided with a dataset called `schools.csv`, which is previewed below.

You have been tasked with answering three key questions about New York City (NYC) public school SAT performance.

```python
# Filter schools where average math score is at least 80% of the max score (800)
# 0.8 * 800 = 640 → we are selecting schools with math scores ≥ 640
best_math_schools = schools[schools['average_math'] >= (0.8 * 800)][['school_name', 'average_math']]

# Sort the filtered schools in descending order to show the highest math scores first
best_math_schools = best_math_schools.sort_values(by='average_math', ascending=False)
```

```python
# Create a new column 'total_SAT' by summing the three SAT section scores:
# Math + Reading + Writing (each out of 800, so max total = 2400)
schools['total_SAT'] = (
    schools['average_math'] 
    + schools['average_reading'] 
    + schools['average_writing']
)

# Sort all schools in descending order of total SAT score
# This brings the highest-performing schools to the top
schools = schools.sort_values(by='total_SAT', ascending=False)

# Select only the school name and total SAT score for the top 10 schools
top_10_schools = schools[["school_name", "total_SAT"]].head(10)

# Display the top 10 schools (first few rows shown with .head())
top_10_schools.head()
```

```python
# Group the dataset by borough and calculate:
# 1. num_schools  → how many schools are in each borough
# 2. average_SAT  → mean total SAT score for each borough
# 3. std_SAT      → standard deviation of total SAT scores (how spread out scores are)
# The .round(2) rounds all numeric results to 2 decimal places
largest_std_dev = schools.groupby('borough').agg(
    num_schools=('school_name', 'count'),
    average_SAT=('total_SAT', 'mean'),
    std_SAT=('total_SAT', 'std')
).round(2)

# Find the maximum standard deviation value among all boroughs
# This tells us which borough has the widest score variation
largest_std_dev_value = largest_std_dev['std_SAT'].max()

# Filter the table to show only the borough with the highest SAT score standard deviation
largest_std_dev = largest_std_dev[largest_std_dev['std_SAT'] == largest_std_dev_value]

# Display the borough with the largest variation in SAT scores
largest_std_dev



```
